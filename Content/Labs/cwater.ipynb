````python
T_b_water = 15.9 
T_e_water = 16.9
u = 9.7
I = 1.0 
m_water = 0.473

P = u * I

t_e = 204 #s

Q = P * t_e

print("aantal joule toegevoegd aan water om 1 graden op te warmen is:", Q , "Joule")

C_water = Q / (m_water * (T_e_water - T_b_water))
print("De warmtecapaciteit van water is:", C_water, "J/(kg*C)")
````
